# Share a plot

A plot in a notebook needs the notebook's Python to draw. To show it to someone else, save
it as a page or as a picture. Docs: <https://mluttikh.github.io/anytimeline/how-to/share.html>

```python
from anytimeline import Timeline, save_html
from anytimeline.synthetic import service_log, traces

plot = Timeline(title="Requests through services",
                lane_order=["gateway", "auth", "api", "cache", "db"])
plot.events(service_log(), time="time", lane="service", color="level", symbol="event",
            tooltip=["request", "user"])
plot
```

## As a page

`save_html` writes one file with the data, the drawing code and the styles: it opens in any
browser without Python or a network, still interactive (pan, zoom, hover with the tooltip
columns, Time / Order, full screen). This writes `requests.html` next to the notebook.

```python
plot.save_html("requests.html")
```

Several plots go in one page, one above the other; linked plots stay linked:

```python
tr = traces(6)
spans = Timeline(title="Six requests", height=300).spans(tr.spans, color="service", expand=1)
save_html("report.html", plot, spans, title="A report")
```

## As a picture

The toolbar's **Save as PNG** button downloads the plot as it is on screen, with its title
and legend. From Python, `save_png` does the same into a file of your choice, also where
the notebook blocks downloads (VS Code). The browser draws the picture, so the plot must be
on screen (run the cell above that shows it first); the file appears right after this cell
has run.

```python
plot.save_png("requests.png")
```
