# Story timelines

For a few dozen events whose text matters (an incident, a release history, a project's
history): each event on a line, its text in a box above or below it, joined by a stem. The
boxes take as many levels as they need so none overlap, at every zoom. Docs:
<https://mluttikh.github.io/anytimeline/how-to/story.html>

```python
from anytimeline import story
from anytimeline.synthetic import incident

events = incident()  # what happened during an incident, as a postmortem lists it
kinds = {"deploy": "#4c78a8", "alert": "#e45756", "page": "#f58518",
         "action": "#54a24b", "status": "#9d755d"}
```

## An incident

A line break in the text template gives a box two lines: the time, then the title. Zoom in:
the boxes settle near the line, on both sides of it. Switch to **Order** in the toolbar to
give each event the same width.

```python
story(events, "{time:%H:%M}\n{title}", color="kind", palette=kinds,
      importance="importance", tooltip=["team"], title="Checkout incident")
```

## A story line per team

```python
story(events, "{time:%H:%M}\n{title}", lane="team", color="kind", palette=kinds,
      importance="importance", title="Checkout incident by team", height=640)
```

## The first half hour

Zoomed in, the boxes need fewer levels: they settle close to the line, above and below it.
`zoom_to` shows a time range from Python (or drag on the time axis).

```python
import numpy as np

plot = story(events, "{time:%H:%M}\n{title}", color="kind", palette=kinds,
             importance="importance", title="The first half hour")
plot.zoom_to(np.datetime64("2026-10-12T13:58"), np.datetime64("2026-10-12T14:30"))
plot
```
